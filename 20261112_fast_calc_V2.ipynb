## はじめに｜計算のやり方を常に工夫しよう

中学受験の算数では、計算を速く正確にすることが大切です。

しかし、難しい筆算を何度も練習するだけでは、計算が速くなるとは限りません。

例えば、次の計算を考えてみましょう。

```text
25 × 48
```

筆算を使っても答えは出せますが、48を4 × 12に分けると、もっと簡単になります。

```text
25 × 48
= 25 × 4 × 12
= 100 × 12
= 1200
```

このように、**計算する前に数をよく見て、もっと簡単な方法がないか考えること**が、高速計算の基本です。

この記事では、次の4つを目標にします。

1. 計算する前に、必ず計算方法を工夫する。
2. 数の分解・約分・計算順序の変更によって筆算を減らす。
3. 計算したら、できる限り別の方法で検算する。
4. 表は眺めるだけで終わらず、必ず自分の手で計算して試す。

特に大切なのは、**計算の答えを丸暗記するのではなく、答えを簡単に出す方法を身につけること**です。

# 第1条　いきなり筆算するな。まず計算の工夫を考えよう

計算式を見たら、すぐに計算を始めるのではなく、数の組み合わせを観察しましょう。

### 例1　25 × 48

```text
25 × 48
= 25 × 4 × 12
= 100 × 12
= 1200
```

25と4を先にかけると100になります。

### 例2　99 × 37

```text
99 × 37
= (100 − 1) × 37
= 3700 − 37
= 3663
```

99を100 − 1と考えれば、2けた同士の筆算をしなくても計算できます。

### 例3　37 × 48 + 37 × 52

```text
37 × 48 + 37 × 52
= 37 × (48 + 52)
= 37 × 100
= 3700
```

37が2回出てきます。

このようなときは、37をまとめると計算回数を減らせます。

### 計算する前に確認する5つのこと

- 10・100・1000を作れないか。
- 数を分解できないか。
- 先に約分できないか。
- 同じ数をまとめられないか。
- 筆算より簡単で、間違えにくい方法がないか。

**計算を速くする一番の方法は、計算する量を減らすことです。**

# 第2条　さくらんぼ算で10・100・1000を作ろう。引き算では±1を利用しよう

さくらんぼ算とは、数をいくつかに分けて計算しやすい数を作る方法です。

## 2.1　足し算では10・100・1000を作る

### 例1　8 + 7

7を2と5に分けます。

```text
8 + 7
= 8 + 2 + 5
= 10 + 5
= 15
```

### 例2　38 + 27

```text
38 + 27
= 38 + 2 + 25
= 40 + 25
= 65
```

### 例3　398 + 76

398を400にすると簡単になります。

```text
398 + 76
= 400 + 76 − 2
= 474
```

400にするために2を足したので、最後に2を引きます。

## 2.2　引き算では100・1000の前後に注目する

### 例1　1001 − 3

1001は1000より1大きい数です。

```text
1001 − 3
= 1000 + 1 − 3
= 1000 − 2
= 998
```

999を使っても計算できます。

```text
1001 − 3
= 999 − 3 + 2
= 998
```

1001を999に変えると2小さくなるので、最後に2を足します。

この場合は、1000 − 2と考えるほうが簡単です。

### 例2　1002 − 7

```text
1002 − 7
= 1000 + 2 − 7
= 1000 − 5
= 995
```

### 例3　10001 − 8

```text
10001 − 8
= 10000 + 1 − 8
= 10000 − 7
= 9993
```

このような計算では、繰り下がりの多い筆算を避けられます。

## 2.3　99・999を引くときは1を戻す

### 例4　723 − 99

```text
723 − 99
= 723 − 100 + 1
= 624
```

99ではなく100を引いたので、最後に1を足します。

### 例5　1500 − 999

```text
1500 − 999
= 1500 − 1000 + 1
= 501
```

## 2.4　397などは400を使う

### 例6　803 − 397

```text
803 − 397
= 803 − 400 + 3
= 406
```

別の方法もあります。

```text
803 − 397
= 806 − 400
= 406
```

803と397の両方に3を足すと、引き算の答えは変わりません。

### 実際に計算してみよう

| 問題 | 計算の工夫 |
|---|---|
| 1001 − 6 | 1000を利用する |
| 1002 − 8 | 1000を利用する |
| 823 − 99 | 100を引いて1戻す |
| 1200 − 998 | 1000を引いて2戻す |
| 703 − 397 | 400を利用する |

答えは、995、994、724、202、306です。

答えを見る前に自分で計算し、足し算で検算しましょう。

# 第3条　かけ算でも10・100・1000・10000を作ろう

かけ算では、数字を分解して計算しやすい組み合わせを探します。

次の表を使って、実際に計算してみましょう。

| 計算 | 答え |
|---|---:|
| 2 × 5 | 10 |
| 4 × 25 | 100 |
| 8 × 125 | 1000 |
| 16 × 625 | 10000 |

表を見て終わりにせず、答えを隠して自分で計算してください。

### 例1　25 × 36

```text
25 × 36
= 25 × 4 × 9
= 100 × 9
= 900
```

### 例2　125 × 48

```text
125 × 48
= 125 × 8 × 6
= 1000 × 6
= 6000
```

### 例3　625 × 32

```text
625 × 32
= 625 × 16 × 2
= 10000 × 2
= 20000
```

### 例4　75 × 16

```text
75 × 16
= 75 × 4 × 4
= 300 × 4
= 1200
```

### 例5　48 × 25

```text
48 × 25
= 12 × 4 × 25
= 12 × 100
= 1200
```

### 練習のポイント

25を見たら4との組み合わせ、125を見たら8との組み合わせを探しましょう。

ただし、どんな数でも無理に100や1000を作る必要はありません。

**最も計算が簡単になる組み合わせを探すことが大切です。**

# 第4条　1から100までの素因数分解を練習し、計算に利用しよう

素因数分解とは、整数を素数だけのかけ算に分けることです。

例えば、12は次のように分解できます。

```text
12 = 2 × 2 × 3
```

同じ数を繰り返しかけるときは、短く書けます。

```text
12 = 2² × 3

24 = 2³ × 3

36 = 2² × 3²
```

2²は2 × 2、2³は2 × 2 × 2という意味です。

なお、1は素数でも合成数でもありません。

## 4.1　素因数分解を使ってかけ算する

### 例1　84 × 25

```text
84 × 25
= 2 × 2 × 3 × 7 × 25
= 4 × 25 × 21
= 100 × 21
= 2100
```

84を分解すると、25と4を組み合わせられます。

## 4.2　素因数分解を使って割り算する

### 例2　96 ÷ 24

```text
96 = 2⁵ × 3

24 = 2³ × 3
```

共通する部分を取り除きます。

```text
96 ÷ 24
= (2⁵ × 3) ÷ (2³ × 3)
= 2²
= 4
```

## 4.3　数字を見て、使える組み合わせを探す

例えば48を見たら、

```text
48 = 6 × 8

48 = 4 × 12

48 = 3 × 16
```

と分けられることに気づくと便利です。

125 × 48なら、48 = 6 × 8を使うと簡単です。

すべての計算で素因数分解する必要はありません。

**1から100までの整数を実際に素因数分解し、計算に使えるように練習しましょう。**

# 第5条　かけ算は順番を変え、数を分け、同じ数をまとめよう

かけ算には、計算する順番を変えても答えが変わらないという性質があります。

## 5.1　順番を変える

```text
25 × 17 × 4
= 25 × 4 × 17
= 100 × 17
= 1700
```

## 5.2　同じ数をまとめる

```text
37 × 48 + 37 × 52
= 37 × (48 + 52)
= 37 × 100
= 3700
```

これを分配法則の利用といいます。

## 5.3　100に近い数を利用する

### 例1　99 × 47

```text
99 × 47
= 100 × 47 − 47
= 4653
```

### 例2　101 × 48

```text
101 × 48
= 100 × 48 + 48
= 4848
```

## 5.4　平方数を利用する

平方数とは、同じ数を2回かけてできる数です。

```text
5² = 5 × 5 = 25

10² = 10 × 10 = 100

20² = 20 × 20 = 400
```

例えば、

```text
53 × 47
= (50 + 3) × (50 − 3)
= 50² − 3²
= 2500 − 9
= 2491
```

50より3大きい数と3小さい数をかけるときは、このような工夫ができます。

**公式を無理に暗記するより、実際に計算して、なぜ簡単になるのか確かめましょう。**

# 第6条　割り算は約分・分解・逆数で効率よく計算しよう

割り算は、いきなり筆算せず、割る数を簡単な形に変えられないか考えましょう。

## 6.1　割る数を分解する

```text
720 ÷ 24
= 720 ÷ 6 ÷ 4
= 120 ÷ 4
= 30
```

24 = 6 × 4なので、6で割ってから4で割れます。

## 6.2　両方を同じ数で割る

```text
840 ÷ 28
= 210 ÷ 7
= 30
```

840と28の両方を4で割ると簡単になります。

## 6.3　25で割るときは4倍する

```text
900 ÷ 25
= 900 × 4 ÷ 100
= 3600 ÷ 100
= 36
```

## 6.4　125で割るときは8倍する

```text
3000 ÷ 125
= 3000 × 8 ÷ 1000
= 24
```

## 6.5　小数で割る場合

```text
350 ÷ 0.25
= 350 × 4
= 1400
```

0.25は1/4なので、0.25で割ることは4倍することと同じです。

## 6.6　分数の割り算

```text
3/4 ÷ 3/8
= 3/4 × 8/3
= 2
```

割る数を逆数にして、かけ算に変えます。

### 実際に試すための表

| 割り算 | 計算の工夫 | 練習問題 |
|---|---|---|
| ÷ 5 | 2倍して10で割る | 370 ÷ 5 |
| ÷ 25 | 4倍して100で割る | 900 ÷ 25 |
| ÷ 125 | 8倍して1000で割る | 3000 ÷ 125 |
| ÷ 0.5 | 2倍する | 38 ÷ 0.5 |
| ÷ 0.25 | 4倍する | 35 ÷ 0.25 |
| ÷ 0.125 | 8倍する | 15 ÷ 0.125 |

答えは順に74、36、24、76、140、120です。

表の計算方法を実際に試し、かけ算に戻して検算しましょう。

# 第7条　小数・分数・割合は、最も計算しやすい形に変えよう

小数・分数・割合は、同じ大きさを違う形で表したものです。

例えば、

```text
0.5 = 1/2 = 50%
```

どの形でも同じ大きさです。

## 7.1　変換を実際に確かめよう

| 小数 | 分数 | 割合 |
|---|---|---|
| 0.5 | 1/2 | 50% |
| 0.25 | 1/4 | 25% |
| 0.125 | 1/8 | 12.5% |
| 0.2 | 1/5 | 20% |
| 0.375 | 3/8 | 37.5% |
| 0.625 | 5/8 | 62.5% |
| 0.75 | 3/4 | 75% |
| 0.04 | 1/25 | 4% |

この表も、ただ眺めるだけではなく、実際に割り算して確かめましょう。

例えば、

```text
3 ÷ 8 = 0.375

5 ÷ 8 = 0.625

1 ÷ 4 = 0.25
```

## 7.2　分数に変えると簡単になる例

```text
64 × 0.375
= 64 × 3/8
= 8 × 3
= 24
```

## 7.3　割合も分数にする

```text
240 × 12.5%
= 240 × 1/8
= 30
```

## 7.4　分数は先に約分する

```text
18/35 × 14/27
= 2/5 × 2/3
= 4/15
```

## 7.5　100%から引く方法も考える

```text
240 × 95%
= 240 × (100% − 5%)
= 240 − 12
= 228
```

割合の計算でも、必ず最も簡単な形を探しましょう。

# 第8条　円周率3.14は最後にまとめて計算しよう

円周率3.14は、最初から何度もかけると計算が大変になります。

できるだけ最後まで3.14を残し、先に数字を整理しましょう。

## 8.1　同じ3.14をまとめる

```text
3.14 × 16 + 3.14 × 9
= 3.14 × (16 + 9)
= 3.14 × 25
= 78.5
```

## 8.2　計算する前に0になる部分を探す

```text
3.14 × 9 + 3.14 × 16 − 3.14 × 25

= 3.14 × (9 + 16 − 25)

= 3.14 × 0

= 0
```

## 8.3　円の面積の差もまとめる

半径10cmの円から半径6cmの円を引いた面積です。

```text
(10² − 6²) × 3.14
= (100 − 36) × 3.14
= 64 × 3.14
= 200.96cm²
```

## 8.4　おうぎ形では先に約分する

半径6cm、中心角60度のおうぎ形の弧の長さです。

```text
6 × 2 × 3.14 × 60/360

= 12 × 3.14 ÷ 6

= 2 × 3.14

= 6.28cm
```

# 第9条　A = BならB = A。逆算を使って検算しよう

A = Bなら、B = Aも成り立ちます。

計算の答えを出したら、反対の計算で確かめられないか考えましょう。

## 9.1　かけ算から割り算を考える

```text
25 × 4 = 100

100 ÷ 25 = 4

100 ÷ 4 = 25
```

```text
125 × 8 = 1000

1000 ÷ 125 = 8

1000 ÷ 8 = 125
```

## 9.2　引き算は足し算で検算する

```text
1001 − 3 = 998

検算

998 + 3 = 1001
```

## 9.3　割り算はかけ算で検算する

```text
840 ÷ 28 = 30

検算

30 × 28 = 840
```

## 9.4　あまりのある割り算

```text
347 ÷ 12 = 28 あまり11
```

検算

```text
12 × 28 + 11
= 336 + 11
= 347
```

さらに、あまりの11が割る数の12より小さいことも確認します。

## 9.5　およその答えも考える

```text
47 × 38 = 1786
```

50 × 40 = 2000なので、1786は大きさとして不自然ではありません。

また、

```text
7 × 8 = 56
```

なので、答えの一の位は6になるはずです。

ただし、概算や一の位が合っていても、必ず正しいとは限りません。

必要に応じて逆算や別の方法で確認しましょう。

**検算は、同じ計算を繰り返すだけでなく、違う方法でも答えが合うか試すことが大切です。**

# 第10条　表を覚えろと言われたら、必ず自分の手で計算して試そう

計算の表を作っても、それを眺めるだけでは計算力は身につきません。

**表の答えを隠し、実際に計算して、検算までやってみましょう。**

## 10.1　まず自力で計算する

例1

```text
25 × 48
= 25 × 4 × 12
= 1200
```

検算

```text
1200 ÷ 25 = 48
```

例2

```text
125 × 64
= 125 × 8 × 8
= 8000
```

検算

```text
8000 ÷ 125 = 64
```

例3

```text
99 × 47
= 100 × 47 − 47
= 4653
```

検算

```text
4653 + 47 = 4700
```

例4

```text
1001 − 3
= 1000 − 2
= 998
```

検算

```text
998 + 3 = 1001
```

## 10.2　計算のやり方を改善する練習

同じ問題を解くときも、毎回ただ答えを出すのではなく、よりよい方法がないか考えましょう。

1回目：答えを隠して、自分で計算する。

2回目：数を分解し、計算回数を減らせないか考える。

3回目：できる限り筆算を使わずに解く。

4回目：逆算・概算・別の方法で検算する。

5回目：間違えた原因を調べて、もう一度解く。

6回目：翌日、何も見ずに再計算する。

## 10.3　速さだけでなく正確さも確認する

次の3つを確認しましょう。

- 正しい答えが出せたか。
- 筆算を減らす工夫ができたか。
- 別の方法で検算できたか。

初めは計算時間よりも正確さを優先します。

正しく計算できるようになってから、少しずつ時間を短くしましょう。

## 7. 計算したら必ず検算しよう

計算を工夫することと、正しく計算することは別です。

例えば、625 × 48を計算するとき、

```
625 × 48
= 625 × 16 × 3
= 10000 × 3
= 30000
```

検算では、反対の計算を利用します。

```
30000 ÷ 625
= 48
```

ただし、毎回わり算で検算すると時間がかかる場合もあります。元の式に戻して別の方法で確かめたり、答えの大きさを見積もったりすることも有効です。

# おわりに｜計算のやり方を工夫する習慣をつけよう

中学受験算数で大切なのは、筆算を速く書くことだけではありません。

計算する前に数を観察し、もっと簡単な方法がないか考えることが重要です。

1001 − 3なら、1000 − 2。

25 × 48なら、25 × 4 × 12。

125 × 64なら、125 × 8 × 8。

3.14が何度も出てきたら、最後にまとめて計算する。

このように、数字の組み合わせや計算順序を工夫すれば、筆算を減らせます。

そして、計算したら必ず検算する習慣をつけましょう。

**計算の表を覚えてくださいと言われても、ただ眺めて暗記するだけで終わらせず、必ず自分の手で計算して試してください。**

そのうえで、もっと速く、もっと間違えにくい方法がないか考えてみましょう。

目標は、筆算を完全になくすことではありません。

**常に計算のやり方を工夫し、最小限の計算で、速く正確に答えを出せるようになることです。**
